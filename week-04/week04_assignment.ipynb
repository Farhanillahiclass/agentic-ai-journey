```python
# question 5 - total with *numbersdef total(*numbers):    """Return the sum of any number of values."""    result = 0    for n in numbers:        result = result + n    return resultprint(total(1, 2))print(total(5, 10, 15, 20))print(total(7))
```

```python
# question 6 - grade functiondef grade(marks):    """Return the letter grade for the given marks."""    if marks >= 80:        return 'A'    elif marks >= 70:        return 'B'    elif marks >= 60:        return 'C'    elif marks >= 50:        return 'D'    else:        return 'F'print('76 marks -> grade', grade(76))print('48 marks -> grade', grade(48))
```

## B · Clean & Correct Code

```python
# question 7 - recursive factorialdef factorial(n):    """Return the factorial of n using recursion."""    if n == 0 or n == 1:        return 1    return n * factorial(n - 1)print('factorial of 5:', factorial(5))print('factorial of 7:', factorial(7))
```

```python
# question 8 - recursive sum of digitsdef sum_of_digits(n):    """Return the sum of the digits of n using recursion."""    if n < 10:        return n    return n % 10 + sum_of_digits(n // 10)print('sum of digits of 472 :', sum_of_digits(472))print('sum of digits of 9999:', sum_of_digits(9999))
```
